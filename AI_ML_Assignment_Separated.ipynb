## Q.1 - Normal Distribution, Histogram and KDE

### Point 1 - Generate a normally distributed dataset using numpy.random.normal()

```python
import numpy as np

# Generate normally distributed dataset
np.random.seed(42)
data = np.random.normal(loc=0, scale=1, size=1000)  # mean=0, std=1, 1000 points

print("Normally distributed dataset generated using numpy.random.normal()")
print(f"Total Points : {len(data)}")
print(f"Mean         : {np.mean(data):.4f}")
print(f"Std Dev      : {np.std(data):.4f}")
print(f"First 10 values: {data[:10]}")
```

### Point 2 - Plot a histogram and KDE using seaborn.histplot() or sns.kdeplot()

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram using seaborn histplot
plt.figure(figsize=(9, 5))
sns.histplot(data, bins=30, color='steelblue', label='Histogram')
plt.title('Histogram of Normally Distributed Data')
plt.xlabel('Value')
plt.ylabel('Count')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# KDE using sns.kdeplot
plt.figure(figsize=(9, 5))
sns.kdeplot(data, color='darkorange', linewidth=2.5, label='KDE Plot')
plt.title('KDE Plot of Normally Distributed Data')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Point 3 - Overlay the mean and standard deviation lines on the plot

```python
mean_val = np.mean(data)
std_val = np.std(data)

plt.figure(figsize=(9, 5))
sns.histplot(data, bins=30, kde=True, color='steelblue', stat='density', label='Histogram + KDE')

# Mean line
plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean = {mean_val:.2f}')

# Std Dev lines
plt.axvline(mean_val + std_val, color='green', linestyle='--', linewidth=1.8, label=f'+1 Std = {mean_val+std_val:.2f}')
plt.axvline(mean_val - std_val, color='green', linestyle='--', linewidth=1.8, label=f'-1 Std = {mean_val-std_val:.2f}')

plt.title('Histogram with Mean and Std Dev Lines')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Mean = {mean_val:.4f}")
print(f"Standard Deviation = {std_val:.4f}")
```

### Point 4 - Use scipy.stats.norm.pdf() to show the bell curve (optional)

```python
from scipy import stats

x = np.linspace(data.min(), data.max(), 300)
bell_curve = stats.norm.pdf(x, mean_val, std_val)

plt.figure(figsize=(9, 5))
sns.histplot(data, bins=30, stat='density', color='skyblue', label='Histogram')
plt.plot(x, bell_curve, color='red', linewidth=2.5, label='Bell Curve (scipy.stats.norm.pdf)')

plt.title('Bell Curve using scipy.stats.norm.pdf()')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Q.2 - Percentiles and Quartiles

### Point 1 - Load a sample dataset and select a numeric column

```python
import pandas as pd
import seaborn as sns

# Load sample dataset
df = sns.load_dataset('tips')
print("Dataset: Tips")
print(df.head())
print(f"\nShape: {df.shape}")
print(f"\nNumeric columns: {df.select_dtypes(include='number').columns.tolist()}")

# Select numeric column
column = df['total_bill']
print(f"\nSelected column: total_bill")
```

### Point 2 - Calculate using numpy.percentile() or df.quantile() - 25th percentile

```python
import numpy as np

# 25th Percentile (Q1)
Q1_numpy = np.percentile(column, 25)
Q1_pandas = df['total_bill'].quantile(0.25)

print("=== 25th Percentile (Q1) ===")
print(f"Using numpy.percentile() : {Q1_numpy:.2f}")
print(f"Using df.quantile(0.25)  : {Q1_pandas:.2f}")
```

### Point 3 - Calculate 50th percentile (median)

```python
# 50th Percentile (Median)
Q2_numpy = np.percentile(column, 50)
Q2_pandas = df['total_bill'].quantile(0.50)

print("=== 50th Percentile - Median (Q2) ===")
print(f"Using numpy.percentile() : {Q2_numpy:.2f}")
print(f"Using df.quantile(0.50)  : {Q2_pandas:.2f}")
```

### Point 4 - Calculate 75th percentile

```python
# 75th Percentile (Q3)
Q3_numpy = np.percentile(column, 75)
Q3_pandas = df['total_bill'].quantile(0.75)

print("=== 75th Percentile (Q3) ===")
print(f"Using numpy.percentile() : {Q3_numpy:.2f}")
print(f"Using df.quantile(0.75)  : {Q3_pandas:.2f}")
```

### Point 5 - Plot these quartiles using vertical lines on a boxplot or histogram

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

sns.histplot(column, kde=True, color='skyblue', bins=25, label='Total Bill')

plt.axvline(Q1_numpy, color='blue', linestyle='--', linewidth=2, label=f'Q1 (25th) = {Q1_numpy:.2f}')
plt.axvline(Q2_numpy, color='red', linestyle='--', linewidth=2, label=f'Q2 / Median (50th) = {Q2_numpy:.2f}')
plt.axvline(Q3_numpy, color='green', linestyle='--', linewidth=2, label=f'Q3 (75th) = {Q3_numpy:.2f}')

plt.title('Histogram with Quartile Vertical Lines')
plt.xlabel('Total Bill ($)')
plt.ylabel('Count')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Q.3 - Boxplot, Five-Number Summary & Outliers

### Point 1 - Print the five-number summary (min, Q1, median, Q3, max) of a column

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset('tips')
col = df['total_bill']

min_val    = col.min()
q1_val     = col.quantile(0.25)
median_val = col.median()
q3_val     = col.quantile(0.75)
max_val    = col.max()

print("=== Five-Number Summary for 'total_bill' ===")
print(f"Min    : {min_val:.2f}")
print(f"Q1     : {q1_val:.2f}")
print(f"Median : {median_val:.2f}")
print(f"Q3     : {q3_val:.2f}")
print(f"Max    : {max_val:.2f}")

# Also using describe()
print("\nUsing describe():")
print(col.describe())
```

### Point 2 - Create a boxplot using seaborn.boxplot() or matplotlib

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))
sns.boxplot(x=df['total_bill'], color='lightcoral')
plt.title('Boxplot of Total Bill')
plt.xlabel('Total Bill ($)')
plt.grid(True, alpha=0.3)
plt.show()
```

### Point 3 - Identify and label outliers from the boxplot

```python
import numpy as np

# IQR method to find outliers
IQR          = q3_val - q1_val
lower_fence  = q1_val - 1.5 * IQR
upper_fence  = q3_val + 1.5 * IQR

outliers = col[(col < lower_fence) | (col > upper_fence)]

print(f"IQR          = {IQR:.2f}")
print(f"Lower Fence  = Q1 - 1.5*IQR = {lower_fence:.2f}")
print(f"Upper Fence  = Q3 + 1.5*IQR = {upper_fence:.2f}")
print(f"\nTotal Outliers Found: {len(outliers)}")
print("Outlier values:", outliers.values)

# Boxplot with outliers labeled
plt.figure(figsize=(10, 5))
ax = sns.boxplot(x=df['total_bill'], color='lightcoral', flierprops=dict(marker='o', color='red', markersize=8))

for val in outliers:
    ax.annotate(f'{val:.1f}', xy=(val, 0), xytext=(val, 0.18),
                fontsize=8, color='darkred', ha='center',
                arrowprops=dict(arrowstyle='->', color='red'))

plt.title('Boxplot with Outliers Labeled')
plt.xlabel('Total Bill ($)')
plt.grid(True, alpha=0.3)
plt.show()
```

## Q.4 - Inferential Statistics & Sampling

### Point 1 - In a markdown cell, define inferential statistics and list real-world use cases

**Inferential Statistics** is a branch of statistics where we use data collected from a **sample** to draw conclusions or make predictions about a **larger population**. 

Unlike descriptive statistics which only summarizes the data we already have, inferential statistics allows us to go beyond that and **infer** things we cannot directly observe.

### Real-World Use Cases:

1. **Polling** - A news channel surveys 1000 people to predict election results for millions of voters.
2. **Clinical Trials** - A new medicine is tested on a small group of patients. If it works, doctors conclude it will work for the general population.
3. **A/B Testing** - Two versions of a website are shown to different user groups. Based on sample results, the better version is deployed for all users.
4. **Quality Control** - In a factory, only a sample of products are checked. If sample is good, the whole batch is approved.
5. **Market Research** - A company surveys 500 customers to understand the preferences of all their customers.

### Point 2 - Load a sample population dataset

```python
import seaborn as sns
import pandas as pd

# Load population dataset
population_df = sns.load_dataset('tips')
population = population_df['total_bill']

print("Population Dataset Loaded: Tips")
print(f"Population Size : {len(population)}")
print(f"Population Mean : {population.mean():.4f}")
print(population_df.head())
```

### Point 3 - Take a random sample using df.sample() and compare the sample mean to the population mean

```python
# Take random sample of 30 rows using df.sample()
sample = population_df.sample(n=30, random_state=10)
sample_mean = sample['total_bill'].mean()
population_mean = population.mean()

print("=== Sample vs Population Mean ===")
print(f"Sample Size    : {len(sample)}")
print(f"Sample Mean    : {sample_mean:.4f}")
print(f"Population Mean: {population_mean:.4f}")
print(f"Difference     : {abs(sample_mean - population_mean):.4f}")

if abs(sample_mean - population_mean) < 2:
    print("\nThe sample mean is close to the population mean! Good sample.")
else:
    print("\nThe sample mean differs from population mean.")
```

### Point 4 - Repeat sampling multiple times and visualize the sample mean distribution

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Repeat sampling 500 times
sample_means = []
num_samples  = 500
sample_size  = 30

for i in range(num_samples):
    s = population_df['total_bill'].sample(n=sample_size)
    sample_means.append(s.mean())

sample_means = np.array(sample_means)

print(f"Sampling repeated {num_samples} times, each with sample size = {sample_size}")
print(f"Mean of all sample means : {sample_means.mean():.4f}")
print(f"Population Mean          : {population_mean:.4f}")
print(f"Standard Error           : {sample_means.std():.4f}")

# Visualize sample mean distribution
plt.figure(figsize=(10, 6))

sns.histplot(sample_means, kde=True, color='mediumpurple', bins=30, label='Sample Means')

plt.axvline(population_mean, color='red', linestyle='--', linewidth=2.5,
            label=f'Population Mean = {population_mean:.2f}')

plt.axvline(sample_means.mean(), color='orange', linestyle=':', linewidth=2.5,
            label=f'Mean of Sample Means = {sample_means.mean():.2f}')

plt.title(f'Sample Mean Distribution (500 samples, n=30)\n- Central Limit Theorem', fontsize=13)
plt.xlabel('Sample Mean of Total Bill ($)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nConclusion: As per Central Limit Theorem, the distribution of sample means")
print("follows a normal distribution even if the population is not perfectly normal.")
```
