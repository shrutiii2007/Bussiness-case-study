## Q.1 — What are Statistics? + Real-World Use Cases

### 📌 What is Statistics?

**Statistics** is the science of **collecting, organizing, analyzing, interpreting, and presenting data**.

It helps us:
- Understand patterns in data
- Make predictions
- Take data-driven decisions

There are two main types:
| Type | Description | Example |
|------|-------------|-------|
| **Descriptive Statistics** | Summarizes existing data | Mean salary of employees |
| **Inferential Statistics** | Makes predictions from a sample | Predicting election results from a survey |

---

### 🌍 5 Real-World Use Cases of Statistics in Data Science

1. **Healthcare** — Analyzing patient data to predict disease risk (e.g., diabetes prediction using blood sugar levels)
2. **Finance/Banking** — Detecting fraud by finding unusual transaction patterns using statistical anomaly detection
3. **E-Commerce (Flipkart/Amazon)** — A/B testing to decide which website layout gets more purchases
4. **Marketing** — Customer segmentation using clustering to send targeted ads
5. **Sports Analytics** — IPL team performance analysis using player statistics to predict match outcomes

## Q.4 — Matplotlib / Seaborn Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style("whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Employee Data — Statistical Visualizations", fontsize=16, fontweight='bold')

# ── Chart 1: Histogram — Age Distribution ──
axes[0, 0].hist(df['Age'], bins=5, color='steelblue', edgecolor='white', rwidth=0.85)
axes[0, 0].set_title("📊 Histogram — Age Distribution", fontsize=12)
axes[0, 0].set_xlabel("Age")
axes[0, 0].set_ylabel("Frequency")

# ── Chart 2: Bar Chart — Department Count ──
dept_count = df['Department'].value_counts()
axes[0, 1].bar(dept_count.index, dept_count.values, color=['#FF6B6B','#4ECDC4','#45B7D1','#96CEB4'], edgecolor='white')
axes[0, 1].set_title("📊 Bar Chart — Department Count", fontsize=12)
axes[0, 1].set_xlabel("Department")
axes[0, 1].set_ylabel("Count")

# ── Chart 3: Pie Chart — Department Percentages ──
axes[1, 0].pie(dept_count.values, labels=dept_count.index, autopct='%1.1f%%',
               colors=['#FF6B6B','#4ECDC4','#45B7D1','#96CEB4'], startangle=90)
axes[1, 0].set_title("🥧 Pie Chart — Department %", fontsize=12)

# ── Chart 4: Boxplot — Salary Distribution ──
axes[1, 1].boxplot(df['Salary'], patch_artist=True,
                   boxprops=dict(facecolor='#4ECDC4', color='navy'),
                   medianprops=dict(color='red', linewidth=2))
axes[1, 1].set_title("📦 Boxplot — Salary Distribution", fontsize=12)
axes[1, 1].set_ylabel("Salary (₹)")

plt.tight_layout()
plt.savefig('charts.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Charts saved!")
```

### 📝 Pattern Observations from Each Chart

1. **Histogram (Age)** — Most employees are between 25–30 years old. The age distribution is slightly right-skewed, meaning younger employees are more common.

2. **Bar Chart (Department Count)** — Sales and IT have equal employees (2 each), followed by HR and Finance (2 each). All departments are evenly balanced.

3. **Pie Chart (Department %)** — Each department holds 25% of total workforce — perfect balance across all departments.

4. **Boxplot (Salary)** — Salary ranges from ₹38,000 to ₹62,000. The median is around ₹53,500. No extreme outliers, suggesting a fairly uniform salary structure.

## Q.5 — Population vs Sample Statistics using NumPy

```python
import numpy as np

# ✅ Step 1: Generate 1,000 random test scores (0–100)
np.random.seed(42)  # for reproducibility
population = np.random.randint(0, 101, size=1000)
print(f"✅ Population created: {len(population)} students")
print(f"First 10 scores: {population[:10]}")
```

```python
# ✅ Step 2: Take a random sample of 100 entries
sample = np.random.choice(population, size=100, replace=False)
print(f"✅ Sample taken: {len(sample)} students")
print(f"First 10 sample scores: {sample[:10]}")
```

```python
# ✅ Step 3: Calculate Mean and Standard Deviation

# Population: use ddof=0 (divide by N)
pop_mean = np.mean(population)
pop_std  = np.std(population, ddof=0)

# Sample: use ddof=1 (divide by N-1) — Bessel's correction
samp_mean = np.mean(sample)
samp_std  = np.std(sample, ddof=1)

print("=" * 45)
print("📊 COMPARISON: Population vs Sample")
print("=" * 45)
print(f"{'Metric':<25} {'Population':>10} {'Sample':>10}")
print("-" * 45)
print(f"{'Size':<25} {len(population):>10} {len(sample):>10}")
print(f"{'Mean':<25} {pop_mean:>10.2f} {samp_mean:>10.2f}")
print(f"{'Standard Deviation':<25} {pop_std:>10.2f} {samp_std:>10.2f}")
print("=" * 45)
```

### 📝 Population vs Sample Statistics — Explained

| Concept | Population | Sample |
|---------|-----------|--------|
| **Definition** | Entire group being studied | A subset of the population |
| **Size** | 1000 students | 100 students |
| **Symbol (Mean)** | μ (mu) | x̄ (x-bar) |
| **Symbol (Std Dev)** | σ (sigma) | s |
| **Std Dev Formula** | Divide by **N** (ddof=0) | Divide by **N-1** (ddof=1) |
| **Why N-1 in sample?** | Bessel's correction — makes sample std dev an unbiased estimator |

**🔑 Key Insight:**  
- The **sample mean** is close to (but rarely exactly equal to) the **population mean**  
- We use **N-1** in sample std dev to compensate for the fact that a sample naturally underestimates the spread of the full population  
- In real life, we almost **never have the full population** — so sample statistics are used to **estimate** population parameters
