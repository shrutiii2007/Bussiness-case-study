## Q1

```python
import numpy as np
from collections import Counter

# dice roll 1000 times
dice = np.random.randint(1, 7, 1000)
print(dice)
```

```python
# frequency count karna hai
freq = Counter(dice)
print(freq)
```

```python
# probability nikalo har face ki
for face in sorted(freq):
    print(f"face {face} : {freq[face]/1000}")
```

```python
# coin toss 500 times
# 0 = tails, 1 = heads
coin = np.random.randint(0, 2, 500)

c = Counter(coin)
print("heads:", c[1])
print("tails:", c[0])
```

```python
# probability
print("P(heads) =", c[1]/500)
print("P(tails) =", c[0]/500)
```

```python
# using value_counts
import pandas as pd

coin_series = pd.Series(coin)
print(coin_series.value_counts())  
print()
print(coin_series.value_counts(normalize=True))  # normalize = probability
```

## Q2

```python
from scipy import stats
import numpy as np

# two datasets with slightly different means
np.random.seed(42)
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(53, 10, 100)

print("mean of data1:", data1.mean())
print("mean of data2:", data2.mean())
```

```python
# t-test
t, p = stats.ttest_ind(data1, data2)

print("t-statistic:", t)
print("p-value:", p)
```

**Interpretation:**

null hypothesis = both groups have same mean

if p-value < 0.05 we reject null hypothesis means there is significant difference between two groups

if p-value > 0.05 we fail to reject null hypothesis means no significant difference

here p-value is less than 0.05 so we reject null hypothesis. the two datasets have significantly different means.

## Q3

```python
import numpy as np
from scipy import stats

# scores before and after training
np.random.seed(5)
before = np.random.normal(60, 8, 30)
after = before + np.random.normal(5, 3, 30)  # after training scores are higher

print("avg before:", before.mean())
print("avg after:", after.mean())
```

```python
# paired t-test because same students before and after
t, p = stats.ttest_rel(before, after)

print("t-stat:", t)
print("p-value:", p)

if p < 0.05:
    print("reject null hypothesis - training helped!")
else:
    print("fail to reject - no improvement found")
```

```python
# confidence interval for mean
diff = after - before

se = stats.sem(diff)  # standard error
mean_d = diff.mean()

ci = stats.t.interval(0.95, df=len(diff)-1, loc=mean_d, scale=se)

print("mean difference:", mean_d)
print("95% confidence interval:", ci)
```

significance level alpha = 0.05

p-value < 0.05 so we reject null hypothesis

training program did improve the scores

confidence interval is above 0 which also confirms positive improvement

## Q4

**Type I Error (False Positive)**

when we reject the null hypothesis but it was actually true

basically we said something is wrong but nothing was wrong

example - a person is healthy but doctor says he has disease. that is false positive.

this is controlled by alpha (significance level). if alpha = 0.05 means 5% chance of type 1 error.

**Type II Error (False Negative)**

when we fail to reject null hypothesis but it was actually false

basically we said nothing is wrong but something was wrong

example - a sick person goes to doctor and doctor says he is fine. that is false negative.

this is represented by beta. to reduce type 2 error we increase sample size.

| | H0 is True | H0 is False |
|---|---|---|
| Reject H0 | Type I Error | Correct |
| Fail to Reject H0 | Correct | Type II Error |
